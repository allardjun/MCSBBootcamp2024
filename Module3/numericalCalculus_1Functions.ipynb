## The coffee bean model

A jar holds `N` scoops of coffee. Each day you drink one scoop’s worth
and top the jar back up with decaf, so the fraction of the jar that is
still caffeinated shrinks by a factor of $1 - 1/N$ every day.

```python
N = 10  # number of scoops in each jar
n_max = 2 * N  # max number of days to simulate

x = np.zeros(n_max)  # fraction caffeinated
x[0] = 1.0  # initial fraction caffeinated

for n in range(1, n_max):

    x[n] = (1 - 1 / N) * x[n - 1]

# finished loop through days

days = np.arange(1, n_max + 1)

fig, ax = plt.subplots()
ax.plot(days, x, "-ok")
ax.set_ylabel("Fraction caffeinated")
ax.set_xlabel("Days")
plt.show()
```

Two things worth taking away from a model this small:

- Sometimes simulation gives unexpected results, i.e., our brains are
  not always good simulators.
- Models have assumptions that are wrong.

Note the index shift against the Matlab version. Matlab arrays start at
1 and the loop runs `n = 2:nMax`; Python arrays start at 0 and the same
loop is `range(1, n_max)`. The array holds the same twenty numbers
either way, and `days` above exists so the plot still counts days from
one.

# 1. Functions of a continuous variable

## 1.1 Functions

Let $u(t)$ be the fluorescence signal from a microscope at time $t$,
with $t$ in seconds:

- $u(t) = A \sin(\omega t) + A_0$.
- Functions are black boxes: you feed an input and get an output.
- Two ways to define a function: in the notebook you are working in, and
  in a separate file.

### Define a function in the notebook

Matlab has two different ways of writing a function, and which one you
get depends on where you put it: `u = @(t) ...` for a one-liner inside a
script, and a `function ... end` block for one in an m-file of its own.
Python has a single form, `def`, and it is written the same way in both
places.

```python
def u(t, A, omega, A_0):
    return A * np.sin(omega * t) + A_0


print(u(1.1, 0.1, 2.6, 0.01))  # t=1.1, A=0.1, omega=2.6, A_0=0.01
```

Four positional arguments in a row is exactly the kind of call that is
easy to get wrong, and reading it back gives you no help at all. Python
will let you name them, which the Matlab version cannot:

```python
print(u(t=1.1, A=0.1, omega=2.6, A_0=0.01))
```

More often you want the parameters fixed and only $t$ free, so that `u`
is a function of time alone. In Matlab you do that by writing the values
into a new anonymous function, `u = @(t) A*sin(omega*t)+A_0`. In Python
you give each parameter a default value, and then you simply do not pass
it:

```python
def u(t, A=1.3, omega=2.6, A_0=0.02):
    """Fluorescence signal at time t. A is in fluorescence intensity units, omega in rad/s."""
    return A * np.sin(omega * t) + A_0


print(u(0.1))
```

`u` is now a function of one argument as far as everything below is
concerned, and the parameters are still there when you want them:
`u(0.1, A=2.0)` overrides one and leaves the other two alone.

### 1.1.1 Functions are different from an array of samples

A function on a continuous set cannot be plotted directly — there are
infinitely many $t$ to evaluate. What you actually plot is an array of
samples, `u(t_array)`.

```python
t_array = np.linspace(0, 1.6, 40)
u_array = u(t_array)  # an array of samples of u

fig, ax = plt.subplots()
ax.plot(t_array, u_array)
ax.set_ylabel("u")
ax.set_xlabel("t")
plt.show()
```

Sampling more coarsely is the same operation with fewer points, and the
coarse samples say much less about the shape between them.

```python
t_array_coarse = np.array([0.1, 0.25, 0.5, 0.7])
u_array_coarse = u(t_array_coarse)

fig, ax = plt.subplots()
ax.plot(t_array, u_array)
ax.plot(t_array_coarse, u_array_coarse, "ob")
ax.axhline(0, color="k", linewidth=0.8)  # Matlab's refline(0,0)
ax.set_ylabel("u")
ax.set_xlabel("t")
plt.show()
```
